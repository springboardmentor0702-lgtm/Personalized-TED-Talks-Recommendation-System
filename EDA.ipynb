```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import ast
import re

sns.set_theme()

def get_list(value):
    """Convert list-like values into a Python list."""
    if pd.isna(value):
        return []

    if isinstance(value, list):
        return value

    if isinstance(value, (tuple, set)):
        return list(value)

    text = str(value).strip()

    if not text or text.lower() in ["nan", "none", "null"]:
        return []

    try:
        parsed = ast.literal_eval(text)
        if isinstance(parsed, (list, tuple, set)):
            return list(parsed)
    except (ValueError, SyntaxError):
        pass

    return [item.strip() for item in text.split(",") if item.strip()]


def show_column(column):
    """Check whether a column exists before using it."""
    if column not in df.columns:
        print(f"Column '{column}' is not available.")
        return False
    return True


def get_numeric(column):
    """Convert a column into numeric values."""
    if column not in df.columns:
        return pd.Series(dtype="float64")

    return pd.to_numeric(df[column], errors="coerce")
```

```python
if show_column("related_talks") and show_column("talk_id"):
    talk_ids = set(df["talk_id"].dropna().astype(str))

    def extract_reference_id(item):
        if isinstance(item, dict):
            for key in ["id", "talk_id", "talk"]:
                if key in item:
                    return str(item[key])
        return str(item)

    missing_references = []

    for row_index, references in df["related_talks_list"].items():
        for item in references:
            reference_id = extract_reference_id(item)

            if reference_id not in talk_ids:
                missing_references.append(
                    (row_index, reference_id)
                )

    print("Total related-talk references not found:", len(missing_references))
    print("Examples:", missing_references[:10])
else:
    print("Cannot validate references: related_talks or talk_id is missing.")
```

```python
before = len(df_clean)


hashable_cols = [col for col in df_clean.columns if not df_clean[col].apply(lambda x: isinstance(x, (list, dict, set))).any()]

df_clean = df_clean.drop_duplicates(subset=hashable_cols).copy()

after = len(df_clean)

print("Rows before:", before)
print("Rows after:", after)
print("Duplicate rows removed:", before - after)
```

```python
important_text_columns = [
    "title", "topics", "description",
    "transcript", "about_speakers"
]

for column in important_text_columns:
    if column in df_clean.columns:
        df_clean[column] = (
            df_clean[column]
            .fillna("")
            .astype(str)
            .str.strip()
        )

print("Missing values in selected text columns handled.")
```

```python
def clean_text(value):
    if pd.isna(value):
        return ""

    text = str(value)

    text = re.sub(r"\s+", " ", text)

    return text.strip()


for column in important_text_columns:
    if column in df_clean.columns:
        df_clean[column] = df_clean[column].apply(clean_text)

print("Text formatting standardised.")
```

```python
combined_columns = [
    "title", "topics", "description", "transcript"
]

available_combined_columns = [
    column for column in combined_columns
    if column in df_clean.columns
]

if available_combined_columns:
    df_clean["combined_text"] = (
        df_clean[available_combined_columns]
        .fillna("")
        .astype(str)
        .agg(" ".join, axis=1)
        .str.replace(r"\s+", " ", regex=True)
        .str.strip()
    )

    print("Combined text column created.")
    print("Columns used:", available_combined_columns)

    if "title" in df_clean.columns:
        display(df_clean[["title", "combined_text"]].head(3))
else:
    print("No suitable text columns found.")
```

```python
combined_columns = [
    "title", "topics", "description", "transcript"
]

available_combined_columns = [
    column for column in combined_columns
    if column in df_clean.columns
]

if available_combined_columns:
    df_clean["combined_text"] = (
        df_clean[available_combined_columns]
        .fillna("")
        .astype(str)
        .agg(" ".join, axis=1)
        .str.replace(r"\s+", " ", regex=True)
        .str.strip()
    )

    print("Combined text column created.")
    print("Columns used:", available_combined_columns)

    if "title" in df_clean.columns:
        display(df_clean[["title", "combined_text"]].head(3))
else:
    print("No suitable text columns found.")
```

```python
print("Cleaned dataset shape:", df_clean.shape)

print("\nRemaining missing values:")
display(df_clean.isnull().sum().sort_values(ascending=False).head(15))

hashable_cols = [col for col in df_clean.columns if not df_clean[col].apply(lambda x: isinstance(x, (list, dict, set))).any()]
print("\nExact duplicate rows remaining (on hashable columns only):", df_clean[hashable_cols].duplicated().sum())

if "combined_text" in df_clean.columns:
    print("\nEmpty combined text rows:",
          (df_clean["combined_text"].str.len() == 0).sum())
```

```python
def get_list(value):
    if value is None:
        return []

    if isinstance(value, (list, tuple, set)):
        return list(value)

    if isinstance(value, float) and np.isnan(value):
        return []

    text = str(value).strip()

    if not text or text.lower() in {"nan", "none", "null"}:
        return []

    try:
        parsed = ast.literal_eval(text)

        if isinstance(parsed, (list, tuple, set)):
            return list(parsed)

    except (ValueError, SyntaxError):
        pass

    return [item.strip() for item in text.split(",") if item.strip()]
```
