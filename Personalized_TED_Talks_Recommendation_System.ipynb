```python
# Step 6: Load the English TED Talks Dataset

import pandas as pd
import os

file_path = "/content/ted_dataset/2020-05-01/ted_talks_en.csv"

df = pd.read_csv(file_path)

print("English TED Talks Dataset Loaded Successfully!")
print("-" * 50)

print("Total Rows:", df.shape[0])
print("Total Columns:", df.shape[1])

print("\nColumn Names:")
print(df.columns.tolist())

print("\nFirst 5 Records:")
display(df.head())
```

```python
# Step 7: Understand the Dataset

print("Dataset Information:")
df.info()

print("\nDataset Shape:")
print(df.shape)

print("\nStatistical Summary:")
display(df.describe(include="all").T)
```

```python
# Step 2: Data Cleaning

# Create a copy of the original dataset
ted_df = df.copy()

# Remove duplicate rows
ted_df.drop_duplicates(subset="talk_id", inplace=True)

# Fill missing values in selected columns
ted_df["occupations"] = ted_df["occupations"].fillna("")
ted_df["about_speakers"] = ted_df["about_speakers"].fillna("")
ted_df["all_speakers"] = ted_df["all_speakers"].fillna("")
ted_df["comments"] = ted_df["comments"].fillna(0)

# Convert date columns to datetime format
ted_df["recorded_date"] = pd.to_datetime(
    ted_df["recorded_date"], errors="coerce"
)

ted_df["published_date"] = pd.to_datetime(
    ted_df["published_date"], errors="coerce"
)

# Remove rows with missing essential recommendation data
ted_df.dropna(
    subset=["title", "description", "transcript"],
    inplace=True
)

# Reset index
ted_df.reset_index(drop=True, inplace=True)

# Display cleaning results
print("Data Cleaning Completed Successfully!")
print("-" * 40)
print("Total Records:", len(ted_df))
print("Total Columns:", ted_df.shape[1])
print("Duplicate Talk IDs:", ted_df["talk_id"].duplicated().sum())
print("\nRemaining Missing Values:")
display(ted_df.isnull().sum().sort_values(ascending=False).head(10))
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_style("whitegrid")

# Graph 1: Top 10 Most Viewed TED Talks
top_talks = ted_df.nlargest(10, "views")

plt.figure(figsize=(12, 6))
sns.barplot(data=top_talks, x="views", y="title")
plt.title("Top 10 Most Viewed TED Talks")
plt.xlabel("Total Views")
plt.ylabel("TED Talk Title")
plt.tight_layout()
plt.show()


# Graph 2: Distribution of TED Talk Duration
plt.figure(figsize=(10, 5))
sns.histplot(data=ted_df, x="duration", bins=30, kde=True)
plt.title("Distribution of TED Talk Duration")
plt.xlabel("Duration (Seconds)")
plt.ylabel("Number of Talks")
plt.tight_layout()
plt.show()


# Graph 3: Top 10 TED Talk Speakers
top_speakers = ted_df["speaker_1"].value_counts().head(10)

plt.figure(figsize=(10, 5))
sns.barplot(x=top_speakers.values, y=top_speakers.index)
plt.title("Top 10 Speakers by Number of TED Talks")
plt.xlabel("Number of Talks")
plt.ylabel("Speaker")
plt.tight_layout()
plt.show()


# Graph 4: TED Talks Published Per Year
ted_df["published_year"] = ted_df["published_date"].dt.year

yearly_talks = ted_df["published_year"].value_counts().sort_index()

plt.figure(figsize=(12, 5))
yearly_talks.plot(kind="line", marker="o")
plt.title("TED Talks Published Per Year")
plt.xlabel("Publication Year")
plt.ylabel("Number of Talks")
plt.tight_layout()
plt.show()
```

```python
import ast
from collections import Counter
import matplotlib.pyplot as plt
import seaborn as sns

# Convert topics into a list safely
def parse_topics(value):
    if isinstance(value, list):
        return value
    try:
        result = ast.literal_eval(value)
        return result if isinstance(result, list) else []
    except (ValueError, SyntaxError, TypeError):
        return []

ted_df["topics_list"] = ted_df["topics"].apply(parse_topics)

# Count the most common topics
topic_counts = Counter(
    topic
    for topics in ted_df["topics_list"]
    for topic in topics
)

top_topics = topic_counts.most_common(15)

topic_df = pd.DataFrame(top_topics, columns=["Topic", "Number of Talks"])

plt.figure(figsize=(12, 6))
sns.barplot(data=topic_df, x="Number of Talks", y="Topic")
plt.title("Top 15 Most Common TED Talk Topics")
plt.tight_layout()
plt.show()

display(topic_df)
```

```python
# Analyze relationships between numerical columns
numeric_cols = ["views", "comments", "duration"]

correlation = ted_df[numeric_cols].corr()

plt.figure(figsize=(7, 5))
sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    vmin=-1,
    vmax=1
)
plt.title("Correlation Between Views, Comments and Duration")
plt.tight_layout()
plt.show()

display(correlation)
```
