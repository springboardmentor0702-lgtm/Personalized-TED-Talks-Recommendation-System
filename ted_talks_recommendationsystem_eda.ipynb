```python
# Dataset load
import pandas as pd
import numpy as np
file_path = "/content/ted_talks_en.csv"

df = pd.read_csv(
    file_path,
    engine="python",
    on_bad_lines="warn"
)

print("Dataset loaded!")
print("Shape:", df.shape)
print("Columns:", df.columns.tolist())

display(df.head())
```

```python

print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])
print("\nColumn names:")
print(df.columns.tolist())
print("\nDataset information:")
df.info()
print("\nSample records:")
display(df.head())
```

```python
#Missing value analysis
missing = df.isnull().sum()
print("Missing values in each column:")
display(missing[missing > 0].sort_values(ascending=False))
print("Duplicate rows:", df.duplicated().sum())
print("\nBlank strings by column:")
for col in df.select_dtypes(include=["object", "string"]).columns:
    count = df[col].astype("string").str.strip().eq("").sum()
    if count > 0:
        print(col, ":", int(count))
```

```python

#Cleaned dataset
clean_df = df.copy()
clean_df.columns = (
    clean_df.columns.str.strip().str.lower()
)
text_cols = clean_df.select_dtypes(
    include=["object", "string"]
).columns
for col in text_cols:
    clean_df[col] = clean_df[col].map(
        lambda x: x.strip() if isinstance(x, str) else x
    )
    clean_df[col] = clean_df[col].replace("", np.nan)
clean_df = clean_df.drop_duplicates()
for col in ["occupations", "about_speakers", "all_speakers"]:
    if col in clean_df.columns:
        clean_df[col] = clean_df[col].fillna("Unknown")
if "comments" in clean_df.columns:
    print("Comments column type:", clean_df["comments"].dtype)
    print("Missing comments:", clean_df["comments"].isna().sum())
if "recorded_date" in clean_df.columns:
    print("Missing recorded dates:",
          clean_df["recorded_date"].isna().sum())
print("\nCleaning steps completed!")
print("Original shape:", df.shape)
print("Current shape:", clean_df.shape)
print("\nRemaining missing values:")
display(
    clean_df.isnull().sum()
    .sort_values(ascending=False)
    .to_frame("Missing Values")
)
```

```python

clean_df.to_csv(
    "/content/ted_talks_cleaned.csv",
    index=False
)

print("Cleaned dataset saved successfully!")
```

```python
#Text Preprocessing
import re
from sklearn.feature_extraction.text import ENGLISH_STOP_WORDS
stop_words = set(ENGLISH_STOP_WORDS)
def clean_text(text):
    if pd.isna(text):
        return ""
    text = str(text).lower()
    text = re.sub(r"http\S+|www\.\S+", " ", text)
    text = re.sub(r"[^a-z0-9\s]", " ", text)
    words = text.split()
    words = [word for word in words if word not in stop_words]
    return " ".join(words)
for col in ["title", "description", "transcript"]:
    if col in clean_df.columns:
        clean_df[col + "_clean"] = clean_df[col].apply(clean_text)
print("Text preprocessing completed!")
display(
    clean_df[
        ["title", "title_clean"]
    ].head(10)
)
```

```python
#Topic Analysis
import pandas as pd
import ast
import matplotlib.pyplot as plt
from collections import Counter
def get_topics(value):
    if pd.isna(value):
        return []
    if isinstance(value, list):
        return value
    try:
        result = ast.literal_eval(value)
        if isinstance(result, list):
            return result
    except (ValueError, SyntaxError):
        pass
    if isinstance(value, str):
        return [x.strip() for x in value.split(",") if x.strip()]
    return []
clean_df["topic_list"] = clean_df["topics"].apply(get_topics)
clean_df["topic_count"] = clean_df["topic_list"].apply(len)
print("Average topics per talk:", clean_df["topic_count"].mean())
print("Minimum topics:", clean_df["topic_count"].min())
print("Maximum topics:", clean_df["topic_count"].max())

topic_counts = Counter(
    str(topic).strip()
    for topics in clean_df["topic_list"]
    for topic in topics
    if str(topic).strip()
)

top_topics = pd.Series(topic_counts).sort_values(
    ascending=False
).head(15)

print("\nTop 15 topics:")
display(top_topics.rename("talk_count").to_frame())

top_topics.sort_values().plot(
    kind="barh",
    figsize=(9, 6),
    title="Top 15 TED Talk Topics"
)
plt.xlabel("Number of TED Talks")
plt.ylabel("Topic")
plt.tight_layout()
plt.show()
```

```python
#Speaker Analysis
import pandas as pd
import matplotlib.pyplot as plt
import ast
from collections import Counter

def get_speakers(value):
    if pd.isna(value):
        return []
    if isinstance(value, str):
        try:
            value = ast.literal_eval(value)
        except (ValueError, SyntaxError):
            return [value.strip()] if value.strip() else []
    if isinstance(value, dict):
        return [
            str(name).strip()
            for name in value.values()
            if str(name).strip()
        ]
    if isinstance(value, (list, tuple, set)):
        names = []
        for item in value:
            if isinstance(item, dict):
                names.extend(get_speakers(item))
            elif str(item).strip():
                names.append(str(item).strip())
        return names

    return []
print(
    "Unique speakers in speaker_1:",
    clean_df["speaker_1"].dropna().astype(str).str.strip().nunique()
)
all_speakers = clean_df["all_speakers"].apply(get_speakers)

speaker_counts = Counter(
    name
    for speakers in all_speakers
    for name in speakers
)
print("Unique speakers in all_speakers:", len(speaker_counts))
multiple_talk_speakers = {
    name: count for name, count in speaker_counts.items()
    if count > 1
}
print(
    "Speakers appearing in multiple talks:",
    len(multiple_talk_speakers)
)
top_10_speakers = pd.Series(speaker_counts).sort_values(
    ascending=False
).head(10)
print("\nTop 10 speakers by number of talks:")
display(top_10_speakers.rename("talk_count").to_frame())
top_10_speakers.sort_values().plot(
    kind="barh",
    figsize=(9, 5),
    title="Top 10 TED Speakers by Number of Talks"
)
plt.xlabel("Number of Talks")
plt.ylabel("Speaker")
plt.tight_layout()
plt.show()
```

```python
#Text Analysis
import pandas as pd
import matplotlib.pyplot as plt
text_columns = [
    "title",
    "description",
    "transcript",
    "about_speakers"
]
available_columns = [
    col for col in text_columns if col in clean_df.columns
]
for col in available_columns:
    text = clean_df[col].fillna("").astype(str).str.strip()
    word_count = text.str.split().str.len()

    print(f"\n--- {col.upper()} ANALYSIS ---")
    print("Empty entries:", (text == "").sum())
    print("Minimum word count:", word_count.min())
    print("Average word count:", round(word_count.mean(), 2))
    print("Maximum word count:", word_count.max())
    if col == "title":
        short_limit = 3
        long_limit = 20
    elif col == "description":
        short_limit = 10
        long_limit = 100
    elif col == "transcript":
        short_limit = 50
        long_limit = 10000
    else:
        short_limit = 10
        long_limit = 500
    print(
        f"Unusually short entries (< {short_limit} words):",
        (text.ne("") & (word_count < short_limit)).sum()
    )
    print(
        f"Unusually long entries (> {long_limit} words):",
        (word_count > long_limit).sum()
    )
    clean_df[col + "_word_count"] = word_count
print("\nSample transcript entries:")
display(
    clean_df[["title", "transcript"]]
    .head(5)
)
```

```python
#Language Anaysis
import pandas as pd
import matplotlib.pyplot as plt
import ast
from collections import Counter
def parse_languages(value):
    if isinstance(value, list):
        return value
    if pd.isna(value):
        return []
    try:
        result = ast.literal_eval(value)
        return result if isinstance(result, list) else []
    except (ValueError, SyntaxError):
        return []
languages = clean_df["available_lang"].apply(parse_languages)
language_counts = Counter(
    lang.strip()
    for langs in languages
    for lang in langs
    if isinstance(lang, str) and lang.strip()
)
top10 = pd.Series(language_counts).nlargest(10)

plt.figure(figsize=(10, 5))
bars = plt.bar(top10.index, top10.values)

plt.bar_label(bars, padding=3)
plt.title("Top 10 Languages Available for TED Talks")
plt.xlabel("Language Code")
plt.ylabel("Number of Talks")
plt.xticks(rotation=45)
plt.grid(axis="y", alpha=0.25)
plt.tight_layout()
plt.show()
```

```python
#Event and Occupation Analsis
import pandas as pd
import matplotlib.pyplot as plt
import ast
def clean_event(value):
    if pd.isna(value):
        return None
    value = str(value).strip()
    try:
        parsed = ast.literal_eval(value)
        if isinstance(parsed, dict):
            value = next(iter(parsed.values()), "")
        elif isinstance(parsed, list):
            value = ", ".join(map(str, parsed))
        else:
            value = str(parsed)
    except (ValueError, SyntaxError):
        pass

    value = value.strip().strip("{}[]'\" ")
    return value if value else None

def clean_occupation(value):
    if pd.isna(value):
        return None
    value = str(value).strip()

    try:
        parsed = ast.literal_eval(value)

        if isinstance(parsed, dict):
            value = ", ".join(str(v) for v in parsed.values())
        elif isinstance(parsed, list):
            value = ", ".join(str(v) for v in parsed)
        else:
            value = str(parsed)

    except (ValueError, SyntaxError):
        pass

    value = value.strip().strip("{}[]'\" ")
    return value if value else None

clean_df["event_clean"] = clean_df["event"].apply(clean_event)
clean_df["occupations_clean"] = clean_df["occupations"].apply(
    clean_occupation
)
top_events = (
    clean_df["event_clean"]
    .dropna()
    .value_counts()
    .head(10)
)
print("Top 10 TED Events:")
display(
    top_events.rename_axis("Event")
    .reset_index(name="Number of Talks")
)
plt.figure(figsize=(10, 5))
top_events.sort_values().plot(kind="barh")
plt.title("Top 10 TED Events by Number of Talks")
plt.xlabel("Number of Talks")
plt.ylabel("TED Event")
plt.tight_layout()
plt.show()
top_occupations = (
    clean_df["occupations_clean"]
    .dropna()
    .value_counts()
    .head(10)
)
print("Top 10 Occupations:")
display(
    top_occupations.rename_axis("Occupation")
    .reset_index(name="Number of Talks")
)
plt.figure(figsize=(10, 6))
top_occupations.sort_values().plot(kind="barh")
plt.title("Top 10 Occupations of TED Speakers")
plt.xlabel("Number of Talks")
plt.ylabel("Occupation")
plt.tight_layout()
plt.show()
```

```python
#Duration Analysis
import pandas as pd
import matplotlib.pyplot as plt
clean_df["duration"] = pd.to_numeric(
    clean_df["duration"], errors="coerce"
)
duration_df = clean_df[
    clean_df["duration"].notna() &
    (clean_df["duration"] > 0)
].copy()
duration_df["duration_minutes"] = duration_df["duration"] / 60
print("TED Talk Duration Analysis")
print("Total talks analyzed:", len(duration_df))
print("Average duration (minutes):",
      round(duration_df["duration_minutes"].mean(), 2))
print("Shortest duration (minutes):",
      round(duration_df["duration_minutes"].min(), 2))
print("Longest duration (minutes):",
      round(duration_df["duration_minutes"].max(), 2))
print("Median duration (minutes):",
      round(duration_df["duration_minutes"].median(), 2))
duration_df["duration_category"] = pd.cut(
    duration_df["duration_minutes"],
    bins=[0, 5, 15, float("inf")],
    labels=["Short (under 5 min)",
            "Medium (5–15 min)",
            "Long (over 15 min)"],
    include_lowest=True
)

print("\nTalks by Duration Category:")
display(
    duration_df["duration_category"]
    .value_counts()
    .reindex([
        "Short (under 5 min)",
        "Medium (5–15 min)",
        "Long (over 15 min)"
    ])
    .rename_axis("Duration Category")
    .reset_index(name="Number of Talks")
)

category_counts = duration_df["duration_category"].value_counts().reindex([
    "Short (under 5 min)",
    "Medium (5–15 min)",
    "Long (over 15 min)"
])

plt.figure(figsize=(9, 5))
bars = plt.bar(
    category_counts.index.astype(str),
    category_counts.values
)
plt.title("TED Talks by Duration Category")
plt.xlabel("Duration Category")
plt.ylabel("Number of Talks")
plt.xticks(rotation=15, ha="right")
for bar in bars:
    plt.text(
        bar.get_x() + bar.get_width() / 2,
        bar.get_height(),
        str(int(bar.get_height())),
        ha="center",
        va="bottom"
    )
plt.tight_layout()
plt.show()
```

```python
#Transcript Analysis
import pandas as pd
import matplotlib.pyplot as plt
clean_df["transcript_word_count"] = (
    clean_df["transcript"]
    .fillna("")
    .astype(str)
    .str.split()
    .str.len()
)
print("TED Talk Transcript Analysis")
print("Total talks:", len(clean_df))
print("Average words:",
      round(clean_df["transcript_word_count"].mean(), 2))
print("Shortest transcript:",
      clean_df["transcript_word_count"].min(), "words")
print("Longest transcript:",
      clean_df["transcript_word_count"].max(), "words")
print("Median words:",
      clean_df["transcript_word_count"].median())
clean_df["transcript_category"] = pd.cut(
    clean_df["transcript_word_count"],
    bins=[-1, 1000, 3000, float("inf")],
    labels=["Short", "Medium", "Long"]
)
print("\nTranscript Length Categories:")
display(
    clean_df["transcript_category"]
    .value_counts()
    .reindex(["Short", "Medium", "Long"])
    .rename_axis("Transcript Length")
    .reset_index(name="Number of Talks")
)
counts = clean_df["transcript_category"].value_counts().reindex(
    ["Short", "Medium", "Long"]
)
plt.figure(figsize=(8, 5))
bars = plt.bar(counts.index.astype(str), counts.values)
plt.title("TED Talks by Transcript Length")
plt.xlabel("Transcript Length")
plt.ylabel("Number of Talks")
for bar in bars:
    plt.text(
        bar.get_x() + bar.get_width() / 2,
        bar.get_height(),
        int(bar.get_height()),
        ha="center",
        va="bottom"
    )
plt.tight_layout()
plt.show()
```

```python
#Related Talk Analysis
import ast
import pandas as pd
import matplotlib.pyplot as plt

def parse_related(value):
    if isinstance(value, list):
        return value
    if pd.isna(value):
        return []
    try:
        result = ast.literal_eval(str(value))
        return result if isinstance(result, (list, tuple)) else []
    except (ValueError, SyntaxError):
        return []

related = clean_df["related_talks"].apply(parse_related)

clean_df["related_talks_count"] = related.apply(len)

print("Total talks:", len(clean_df))
print("Talks with related references:", (clean_df["related_talks_count"] > 0).sum())
print("Average related talks:", clean_df["related_talks_count"].mean())
print("Maximum related talks:", clean_df["related_talks_count"].max())

print("Talks with no related references:", (clean_df["related_talks_count"] == 0).sum())
```

```python
#Final data cleaning
import pandas as pd
import re

final_df = clean_df.copy()

final_df.columns = final_df.columns.str.strip().str.lower()

for col in ["title", "topics", "description", "transcript", "about_speakers"]:
    if col in final_df.columns:
        final_df[col] = final_df[col].map(
            lambda x: re.sub(r"\s+", " ", x).strip()
            if isinstance(x, str) else x
        )
before = len(final_df)

final_df = final_df.loc[
    ~final_df.astype(str).duplicated()
].copy()

print("Duplicate rows removed:", before - len(final_df))


text_columns = ["title", "topics", "description", "transcript"]

for col in text_columns:
    if col not in final_df.columns:
        final_df[col] = ""

final_df["combined_text"] = (
    final_df[text_columns]
    .fillna("")
    .astype(str)
    .agg(" ".join, axis=1)
    .str.replace(r"\s+", " ", regex=True)
    .str.strip()
)


important_cols = [
    "title", "topics", "description", "transcript",
    "speaker_1", "views", "duration", "comments"
]
print("\nMissing values:")
print(final_df[[c for c in important_cols if c in final_df.columns]].isna().sum())


print("\nEmpty combined-text rows:",
      (final_df["combined_text"] == "").sum())

final_df.to_csv("/content/ted_talks_eda_final.csv", index=False)
print("\nSaved: /content/ted_talks_eda_final.csv")
print("Final dataset shape:", final_df.shape)
```
