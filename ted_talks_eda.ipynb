```python

# ============================================================
# INFOSYS PROJECT: TED TALKS RECOMMENDATION SYSTEM
# EDA CODE 1: TOPICS, SPEAKERS, TEXT, VIEWS, DURATION,
#             AND CORRELATION ANALYSIS
# This cell only analyzes and visualizes the existing df.
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import ast
import re
from collections import Counter

# Helper: safely interpret list-like values without modifying df
def parse_list(value):
    if isinstance(value, (list, tuple, set)):
        return list(value)
    if pd.isna(value):
        return []
    text = str(value).strip()
    if not text or text.casefold() in {"unknown", "none", "nan", "[]"}:
        return []
    try:
        parsed = ast.literal_eval(text)
        if isinstance(parsed, (list, tuple, set)):
            return list(parsed)
    except (ValueError, SyntaxError):
        pass
    return [x.strip() for x in text.split(",") if x.strip()]

# Helper: count words without modifying the original text
def word_count(value):
    if pd.isna(value):
        return 0
    return len(re.findall(r"\b[\w'-]+\b", str(value)))

# Helper: show a horizontal bar chart of top categories
def bar_chart(counts, title, xlabel, top_n=10):
    counts = counts.head(top_n).sort_values()
    if counts.empty:
        print("No data available for:", title)
        return
    plt.figure(figsize=(10, 6))
    plt.barh(counts.index.astype(str), counts.values)
    plt.title(title)
    plt.xlabel(xlabel)
    plt.ylabel("Category")
    plt.tight_layout()
    plt.show()


# ------------------------------------------------------------
# 1. TOPIC ANALYSIS
# ------------------------------------------------------------
print("\n1. TOPIC ANALYSIS")
print("=" * 55)

topic_lists = df["topics"].apply(parse_list)
topic_lists = topic_lists.apply(
    lambda xs: list(dict.fromkeys(
        str(x).strip().casefold() for x in xs if str(x).strip()
    ))
)

topics_per_talk = topic_lists.apply(len)
topic_counts = Counter(x for xs in topic_lists for x in xs)

print("Talks with topics:", (topics_per_talk > 0).sum())
print("Talks without topics:", (topics_per_talk == 0).sum())
print("Average topics per talk:", round(topics_per_talk.mean(), 2))
print("Maximum topics per talk:", topics_per_talk.max())

print("\nTop 15 topics:")
top_topics = pd.Series(topic_counts).sort_values(ascending=False).head(15)
print(top_topics)

bar_chart(top_topics, "Top 15 TED Talk Topics",
          "Number of Topic Assignments", 15)

topic_distribution = topics_per_talk.value_counts().sort_index()
plt.figure(figsize=(9, 5))
plt.bar(topic_distribution.index.astype(str), topic_distribution.values)
plt.title("Number of Topics per TED Talk")
plt.xlabel("Topics per Talk")
plt.ylabel("Number of TED Talks")
plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# 2. SPEAKER ANALYSIS
# ------------------------------------------------------------
print("\n2. SPEAKER ANALYSIS")
print("=" * 55)

speaker_counts = Counter()

for _, row in df.iterrows():
    speakers = []

    if pd.notna(row.get("speaker_1")):
        speakers.append(str(row["speaker_1"]).strip())

    speakers.extend([
        str(x).strip() for x in parse_list(row.get("all_speakers"))
        if str(x).strip()
    ])

    # Count each speaker at most once per talk
    speakers = set(s.casefold() for s in speakers if s)
    speaker_counts.update(speakers)

print("Unique speakers:", len(speaker_counts))
print("Speakers with multiple talks:",
      sum(n > 1 for n in speaker_counts.values()))

top_speakers = pd.Series(speaker_counts).sort_values(
    ascending=False
).head(10)

print("\nTop 10 speakers:")
print(top_speakers)

bar_chart(top_speakers, "Top 10 Speakers by Talk Count",
          "Number of Talks", 10)


# ------------------------------------------------------------
# 3. TEXT ANALYSIS
# ------------------------------------------------------------
print("\n3. TEXT ANALYSIS")
print("=" * 55)

text_columns = [
    "title", "description", "transcript", "about_speakers"
]
text_results = []

for col in text_columns:
    if col not in df.columns:
        print(f"{col}: column not found")
        continue

    text = df[col].fillna("").astype(str).str.strip()
    lengths = text.apply(word_count)

    print(f"\n{col.upper()}")
    print("Empty entries:", (text == "").sum())
    print("Average word count:", round(lengths.mean(), 2))
    print("Median word count:", lengths.median())
    print("Entries under 5 words:", (lengths < 5).sum())
    print("Entries over 1000 words:", (lengths > 1000).sum())

    text_results.append({
        "Column": col,
        "Empty": int((text == "").sum()),
        "Mean words": round(lengths.mean(), 2),
        "Median words": lengths.median(),
        "Under 5 words": int((lengths < 5).sum()),
        "Over 1000 words": int((lengths > 1000).sum())
    })

print("\nText quality summary:")
print(pd.DataFrame(text_results).to_string(index=False))

# Compare word-count distributions
plt.figure(figsize=(10, 5))
for col in text_columns:
    if col in df.columns:
        lengths = df[col].fillna("").astype(str).apply(word_count)
        plt.hist(lengths, bins=30, alpha=0.4, label=col)

plt.title("Word Count Distributions by Text Column")
plt.xlabel("Word Count")
plt.ylabel("Number of TED Talks")
plt.legend()
plt.tight_layout()
plt.show()

# Inspect transcript samples without printing entire transcripts
if "transcript" in df.columns:
    print("\nTranscript samples (first 250 characters):")
    for value in df["transcript"].dropna().head(3):
        print(str(value)[:250], "\n")


# ------------------------------------------------------------
# 4. VIEWS ANALYSIS
# ------------------------------------------------------------
print("\n4. VIEWS ANALYSIS")
print("=" * 55)

views = pd.to_numeric(df["views"], errors="coerce")
print(views.describe())
print("Missing/non-numeric views:", views.isna().sum())

q1, q3 = views.quantile(0.25), views.quantile(0.75)
iqr = q3 - q1
print("Potential IQR outliers:",
      ((views < q1 - 1.5 * iqr) | (views > q3 + 1.5 * iqr)).sum())

# Top 10 talks by views
top_views = df.loc[views.nlargest(10).index, ["title", "views"]]
print("\nTop 10 most-viewed talks:")
print(top_views.to_string(index=False))

plt.figure(figsize=(11, 6))
plot_views = top_views.copy()
plot_views["views"] = pd.to_numeric(plot_views["views"])
plot_views = plot_views.sort_values("views")
plt.barh(plot_views["title"], plot_views["views"])
plt.title("Top 10 TED Talks by Views")
plt.xlabel("Views")
plt.tight_layout()
plt.show()

plt.figure(figsize=(9, 5))
plt.hist(views.dropna(), bins=30, edgecolor="black")
plt.title("Distribution of TED Talk Views")
plt.xlabel("Views")
plt.ylabel("Number of Talks")
plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# 5. DURATION ANALYSIS
# ------------------------------------------------------------
print("\n5. DURATION ANALYSIS")
print("=" * 55)

duration = pd.to_numeric(df["duration"], errors="coerce")
print(duration.describe())
print("Missing/non-numeric durations:", duration.isna().sum())
print("Zero or negative durations:", (duration <= 0).sum())

q1, q3 = duration.quantile(0.25), duration.quantile(0.75)
iqr = q3 - q1
print("Potential IQR outliers:",
      ((duration < q1 - 1.5 * iqr) |
       (duration > q3 + 1.5 * iqr)).sum())

plt.figure(figsize=(9, 5))
plt.hist(duration.dropna(), bins=30, edgecolor="black")
plt.title("Distribution of TED Talk Duration")
plt.xlabel("Recorded Duration (verify units)")
plt.ylabel("Number of Talks")
plt.tight_layout()
plt.show()

top_duration = df.loc[duration.nlargest(10).index, ["title", "duration"]].copy()
top_duration["duration"] = pd.to_numeric(top_duration["duration"])
top_duration = top_duration.sort_values("duration")

plt.figure(figsize=(11, 6))
plt.barh(top_duration["title"], top_duration["duration"])
plt.title("Top 10 Longest TED Talks")
plt.xlabel("Recorded Duration")
plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# 6. CORRELATION ANALYSIS
# ------------------------------------------------------------
print("\n6. CORRELATION ANALYSIS")
print("=" * 55)

numeric_cols = [
    col for col in ["views", "comments", "duration"]
    if col in df.columns
]
numeric_data = df[numeric_cols].apply(pd.to_numeric, errors="coerce")
corr = numeric_data.corr()

print(corr.round(3))

plt.figure(figsize=(7, 5))
plt.imshow(corr, cmap="coolwarm", vmin=-1, vmax=1)
plt.colorbar(label="Correlation")
plt.xticks(range(len(numeric_cols)), numeric_cols, rotation=30)
plt.yticks(range(len(numeric_cols)), numeric_cols)

for i in range(len(numeric_cols)):
    for j in range(len(numeric_cols)):
        val = corr.iloc[i, j]
        plt.text(j, i, "N/A" if pd.isna(val) else f"{val:.2f}",
                 ha="center", va="center")

plt.title("Correlation Between Numerical Features")
plt.tight_layout()
plt.show()
```

```python

# ============================================================
# EDA CODE 2: DATES, LANGUAGES, EVENTS, OCCUPATIONS,
#             COMMENTS AND RELATED TALKS
# Analysis only: no cleaning or saving.
# ============================================================

import pandas as pd
import matplotlib.pyplot as plt
import ast
from collections import Counter

# Helper: parse list-like fields
def parse_list_eda(value):
    if isinstance(value, (list, tuple, set)):
        return list(value)
    if pd.isna(value):
        return []
    text = str(value).strip()
    if not text or text.casefold() in {"unknown", "none", "nan", "[]"}:
        return []
    try:
        parsed = ast.literal_eval(text)
        if isinstance(parsed, (list, tuple, set)):
            return list(parsed)
    except (ValueError, SyntaxError):
        pass
    return [x.strip() for x in text.split(",") if x.strip()]

# Helper: horizontal bar chart
def plot_counts_eda(counts, title, xlabel, top_n=10):
    counts = counts.head(top_n).sort_values()
    if counts.empty:
        print("No values available for:", title)
        return
    plt.figure(figsize=(10, 6))
    plt.barh(counts.index.astype(str), counts.values)
    plt.title(title)
    plt.xlabel(xlabel)
    plt.tight_layout()
    plt.show()


# ------------------------------------------------------------
# 7. DATE ANALYSIS
# ------------------------------------------------------------
print("\n7. DATE ANALYSIS")
print("=" * 55)

for col in ["recorded_date", "published_date"]:
    if col not in df.columns:
        print(f"{col}: column not found")
        continue

    dates = pd.to_datetime(df[col], errors="coerce")
    print(f"\n{col}")
    print("Missing or invalid dates:", dates.isna().sum())

    if dates.notna().any():
        print("Earliest date:", dates.min())
        print("Latest date:", dates.max())

# Publication-year trend
if "published_date" in df.columns:
    published = pd.to_datetime(df["published_date"], errors="coerce")
    yearly_counts = published.dropna().dt.year.value_counts().sort_index()

    print("\nTalks published each year:")
    print(yearly_counts)

    if not yearly_counts.empty:
        plt.figure(figsize=(11, 5))
        plt.plot(yearly_counts.index, yearly_counts.values, marker="o")
        plt.title("TED Talks Published by Year")
        plt.xlabel("Year")
        plt.ylabel("Number of Talks")
        plt.grid(True, alpha=0.3)
        plt.tight_layout()
        plt.show()


# ------------------------------------------------------------
# 8. LANGUAGE ANALYSIS
# ------------------------------------------------------------
print("\n8. LANGUAGE ANALYSIS")
print("=" * 55)

if "native_lang" in df.columns:
    native = df["native_lang"].fillna("").astype(str).str.strip()
    print("Missing/blank native language:", (native == "").sum())
    native_counts = native[native != ""].value_counts()

    print("\nTop native languages:")
    print(native_counts.head(15))
    plot_counts_eda(native_counts, "Top Native Languages",
                    "Number of Talks", 15)

if "available_lang" in df.columns:
    available = df["available_lang"].apply(parse_list_eda)
    language_counts = Counter(
        str(lang).strip().casefold()
        for langs in available
        for lang in langs
        if str(lang).strip()
    )

    language_series = pd.Series(language_counts).sort_values(ascending=False)

    print("\nTop available translation languages:")
    print(language_series.head(15))
    plot_counts_eda(language_series, "Top Available Languages",
                    "Number of Language Entries", 15)


# ------------------------------------------------------------
# 9. EVENT ANALYSIS
# ------------------------------------------------------------
print("\n9. EVENT ANALYSIS")
print("=" * 55)

if "event" in df.columns:
    events = df["event"].fillna("").astype(str).str.strip()
    print("Missing/blank event values:", (events == "").sum())

    event_counts = events[events != ""].value_counts()
    print("\nTop 15 events:")
    print(event_counts.head(15))

    plot_counts_eda(event_counts, "Top 15 TED Talk Events",
                    "Number of Talks", 15)


# ------------------------------------------------------------
# 10. OCCUPATION ANALYSIS
# ------------------------------------------------------------
print("\n10. OCCUPATION ANALYSIS")
print("=" * 55)

if "occupations" in df.columns:
    occupation_counts = Counter(
        str(occupation).strip().casefold()
        for values in df["occupations"].apply(parse_list_eda)
        for occupation in values
        if str(occupation).strip()
        and str(occupation).strip().casefold() != "unknown"
    )

    occupation_series = pd.Series(occupation_counts).sort_values(
        ascending=False
    )

    print("Top 15 occupations:")
    print(occupation_series.head(15))

    plot_counts_eda(occupation_series, "Top 15 Speaker Occupations",
                    "Number of Occupation Entries", 15)


# ------------------------------------------------------------
# 11. COMMENTS ANALYSIS
# ------------------------------------------------------------
print("\n11. COMMENTS ANALYSIS")
print("=" * 55)

comments = pd.to_numeric(df["comments"], errors="coerce")
print("Comments summary:")
print(comments.describe())
print("Missing/non-numeric comments:", comments.isna().sum())

q1, q3 = comments.quantile(0.25), comments.quantile(0.75)
iqr = q3 - q1

print("Potential IQR outliers:",
      ((comments < q1 - 1.5 * iqr) |
       (comments > q3 + 1.5 * iqr)).sum())

# Top 10 talks by comment count
top_comments_idx = comments.nlargest(10).index
top_comments = df.loc[top_comments_idx, ["title", "comments"]].copy()
top_comments["comments"] = pd.to_numeric(top_comments["comments"])
top_comments = top_comments.sort_values("comments")

print("\nTop 10 talks by comments:")
print(top_comments.sort_values("comments", ascending=False).to_string(index=False))

plt.figure(figsize=(11, 6))
plt.barh(top_comments["title"], top_comments["comments"])
plt.title("Top 10 TED Talks by Comments")
plt.xlabel("Number of Comments")
plt.tight_layout()
plt.show()

plt.figure(figsize=(9, 5))
plt.hist(comments.dropna(), bins=30, edgecolor="black")
plt.title("Distribution of TED Talk Comments")
plt.xlabel("Comments")
plt.ylabel("Number of Talks")
plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# 12. RELATED TALKS ANALYSIS
# ------------------------------------------------------------
print("\n12. RELATED TALKS ANALYSIS")
print("=" * 55)

if "related_talks" in df.columns:
    related_lists = df["related_talks"].apply(parse_list_eda)

    def extract_related_id(item):
        if isinstance(item, dict):
            for key in ["talk_id", "id", "id_talk"]:
                if key in item:
                    return str(item[key]).strip()
            return None
        return str(item).strip()

    related_counts = related_lists.apply(
        lambda items: sum(
            bool(extract_related_id(item)) for item in items
        )
    )

    print("Talks with related-talk entries:", (related_counts > 0).sum())
    print("Talks with no related-talk entries:", (related_counts == 0).sum())
    print("Average related talks per record:", round(related_counts.mean(), 2))
    print("Maximum related talks:", related_counts.max())

    # Check whether references match talk IDs in the current dataset
    if "talk_id" in df.columns:
        known_ids = set(df["talk_id"].dropna().astype(str).str.strip())

        refs = [
            extract_related_id(item)
            for items in related_lists
            for item in items
        ]
        refs = [ref for ref in refs if ref]

        matched = sum(ref in known_ids for ref in refs)
        unmatched = len(refs) - matched

        print("Total related-talk references:", len(refs))
        print("References matching dataset IDs:", matched)
        print("References not found in dataset:", unmatched)
        print("Unmatched references may refer to talks outside this dataset.")

    # Plot the distribution of related-talk counts
    related_distribution = related_counts.value_counts().sort_index()

    plt.figure(figsize=(9, 5))
    plt.bar(related_distribution.index.astype(str),
            related_distribution.values)
    plt.title("Related Talks per TED Talk")
    plt.xlabel("Number of Related Talks")
    plt.ylabel("Number of TED Talks")
    plt.tight_layout()
    plt.show()

else:
    print("The related_talks column was not found.")


# ------------------------------------------------------------
# FINAL EDA CHECKLIST SUMMARY
# ------------------------------------------------------------
print("\nFINAL EDA CHECK")
print("=" * 55)
print("Rows:", len(df))
print("Columns:", len(df.columns))
print("Exact duplicate rows:", df.duplicated().sum())
print("Missing cells:", df.isnull().sum().sum())
print("\nMissing values by column:")
print(df.isnull().sum().sort_values(ascending=False))
```
