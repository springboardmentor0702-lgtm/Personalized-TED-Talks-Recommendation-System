```python
import re
display(HTML("<h2 style='text-align:center;'>Data Cleaning Decisions</h2>"))
#Data Cleaning Decisions
text_columns = [
    "title",
    "topics",
    "description",
    "transcript"
]
for col in text_columns:
    df[col] = df[col].fillna("").astype(str)

#cleaning code
def clean_text(text):
    text = text.lower()
    text = re.sub(r"http\S+|www\.\S+", " ", text)
    text = re.sub(r"[^a-z0-9\s]", " ", text)
    text = re.sub(r"\s+", " ", text)
    return text.strip()

df["occupations"] = df["occupations"].fillna("Unknown")
df["about_speakers"] = df["about_speakers"].fillna("")
df["all_speakers"] = df["all_speakers"].fillna(df["speaker_1"])

#Standardize text
df["title"] = (
    df["title"]
      .str.lower()
      .str.strip()
)
df["description"] = (
    df["description"]
      .str.lower()
      .str.strip()
)
df["transcript"] = (
    df["transcript"]
      .str.lower()
      .str.strip()
)
df["topics"] = (
    df["topics"]
      .str.lower()
      .str.strip()
)

#create combined text
df["combined_text"] = (
    df["title"].fillna("") + " " +
    df["topics"].fillna("") + " " +
    df["description"].fillna("") + " " +
    df["transcript"].fillna("")
)

display(df[["title", "combined_text"]].head(3))

# Apply text cleaning to combined text
df["cleaned_text"] = df["combined_text"].apply(clean_text)

# Display original and cleaned text
display(
    df[["title", "combined_text", "cleaned_text"]].head(5)
)

# Verify cleaned data
print("Total records:", len(df))
print(
    "Empty cleaned records:",
    (df["cleaned_text"].str.strip() == "").sum()
)
```

```python
display(HTML("<h2 style='text-align:center;'>Cleaned text</h2>"))
# Convert topic lists into normal text
df["topics_text"] = df["topic_list"].apply(
    lambda topics: " ".join(topics)
)

# Create improved combined text
df["combined_text"] = (
    df["title"] + " " +
    df["topics_text"] + " " +
    df["description"] + " " +
    df["transcript"]
)

# Apply text cleaning again
df["cleaned_text"] = df["combined_text"].apply(clean_text)

# Display results
display(
    df[["title", "topics_text", "cleaned_text"]].head(5)
)
```

```python
#TF-IDF

# re — helps clean unnecessary characters and extra spaces.
# TfidfVectorizer — converts text into numerical vectors.
# cosine_similarity — measures how similar two TED Talks are.
import re
import pickle
from sklearn.feature_extraction.text import TfidfVectorizer

display(HTML("<h2 style='text-align:center;'>TF- IDF</h2>"))
tfidf = TfidfVectorizer(
    stop_words="english",
    max_features=10000,
    ngram_range=(1, 2),
    min_df=2
)

tfidf_matrix = tfidf.fit_transform(df["cleaned_text"])

print("TF-IDF matrix shape:", tfidf_matrix.shape)
print("Number of TED Talks:", tfidf_matrix.shape[0])
print("Number of Features:", tfidf_matrix.shape[1])


# Save the TF-IDF vectorizer
with open("tfidf_vectorizer.pkl", "wb") as file:
    pickle.dump(tfidf, file)

# Save the TF-IDF matrix
with open("tfidf_matrix.pkl", "wb") as file:
    pickle.dump(tfidf_matrix, file)

print("TF-IDF vectorizer saved successfully!")
print("TF-IDF matrix saved successfully!")

#verify saved files
print("Vectorizer exists:", os.path.exists("tfidf_vectorizer.pkl"))
print("Matrix exists:", os.path.exists("tfidf_matrix.pkl"))
```

```python
display(HTML("<h2 style='text-align:center;'>Cosine Similarity</h2>"))
# cosine_similarity — measures how similar two TED Talks are.
from sklearn.metrics.pairwise import cosine_similarity

# Calculate similarity between TED Talks
cosine_sim = cosine_similarity(tfidf_matrix, tfidf_matrix)

# Display similarity matrix information
print("Cosine Similarity Matrix Shape:", cosine_sim.shape)

# Display similarity scores for the first five talks
display(pd.DataFrame(cosine_sim[:5, :5]))


# Save cosine similarity matrix
with open("cosine_similarity.pkl", "wb") as file:
    pickle.dump(cosine_sim, file)

print("Cosine similarity matrix saved successfully!")
```

```python
display(HTML("<h2 style='text-align:center;'>Recommendation Engine</h2>"))
def recommend_talk(title, top_n=5):

    # Find the selected TED Talk
    matching_talks = df[
        df["title"].str.lower().str.strip() == title.lower().strip()
    ]

    if matching_talks.empty:
        print("TED Talk not found!")
        return

    # Get the index of the selected talk
    talk_index = matching_talks.index[0]

    # Get similarity scores for the selected talk
    similarity_scores = list(enumerate(cosine_sim[talk_index]))

    # Sort talks by similarity score
    similarity_scores = sorted(
        similarity_scores,
        key=lambda x: x[1],
        reverse=True
    )

    # Remove the selected talk itself
    similarity_scores = [
        score for score in similarity_scores
        if score[0] != talk_index
    ]

    # Select the top recommended talks
    top_recommendations = similarity_scores[:top_n]

    # Display recommendations
    recommendations = []

    for index, score in top_recommendations:

        recommendations.append({
            "Title": df.iloc[index]["title"],
            "Speaker": df.iloc[index]["speaker_1"],
            "Views": df.iloc[index]["views"],
            "Similarity Score": round(score, 4),
            "URL": df.iloc[index]["url"]
        })

    return pd.DataFrame(recommendations)
```

```python
# Test recommendation system with multiple TED Talks

test_talks = [
    "Averting the climate crisis",
    "The power of vulnerability",
    "Do schools kill creativity?",
    "How great leaders inspire action"
]

for talk in test_talks:
    print("\nInput TED Talk:", talk)
    print("-" * 60)

    result = recommend_talk(talk, top_n=5)
    display(result)
```

```python
# Evaluate recommendation similarity scores
display(HTML("<h2 style='text-align:center;'>Evaluating Recommendation Similarity Scores</h2>"))
test_talks = [
    "Averting the climate crisis",
    "The power of vulnerability",
    "Do schools kill creativity?",
    "How great leaders inspire action"
]

evaluation_results = []

for talk in test_talks:

    recommendations = recommend_talk(talk, top_n=5)

    if recommendations is not None and not recommendations.empty:

        average_score = recommendations["Similarity Score"].mean()

        evaluation_results.append({
            "Input TED Talk": talk,
            "Average Similarity Score": round(average_score, 4)
        })

evaluation_df = pd.DataFrame(evaluation_results)

display(evaluation_df)

print(
    "Overall Average Similarity Score:",
    round(evaluation_df["Average Similarity Score"].mean(), 4)
)
```

Hybrid TED Talk Recommendation System

We'll combine two signals:
1. Content similarity (80%) – TF-IDF and cosine similarity.
2. Topic similarity (20%) – common topics between two TED Talks.

For example:
- Talk A topics: Climate, Environment, Science
- Talk B topics: Climate, Science, Technology
- Common topics: Climate, Science
- Total unique topics: Climate, Environment, Science, Technology
Topic similarity = 2 / 4 = 0.50



```python
# Calculate topic similarity using Jaccard similarity
display(HTML("<h2 style='text-align:center;'>Hybrid TED Talk Recommendation System</h2>"))
def calculate_topic_similarity(topics1, topics2):

    set1 = set(topics1)
    set2 = set(topics2)

    if not set1 or not set2:
        return 0

    common_topics = set1.intersection(set2)
    total_topics = set1.union(set2)

    return len(common_topics) / len(total_topics)


print("Topic similarity function created successfully!")
```

How the hybrid score works:
\[
Hybrid = (0.8 \times Content) + (0.2 \times Topic)
\]

This gives more importance to TF-IDF content similarity while also considering shared topics.

hybrid model uses:
- TF-IDF vectorizer
- TF-IDF matrix
- Cosine similarity matrix
- Topic lists
- Hybrid weights (80% content + 20% topics)

```python
def recommend_hybrid(title, top_n=5):

    # Find the input TED Talk
    matching_talks = df[
        df["title"].str.lower().str.strip() == title.lower().strip()
    ]

    if matching_talks.empty:
        print("TED Talk not found!")
        return

    talk_position = matching_talks.index[0]

    # Get input talk topics
    input_topics = df.loc[talk_position, "topic_list"]

    recommendations = []

    # Compare with every other TED Talk
    for position in range(len(df)):

        if position == talk_position:
            continue

        candidate_topics = df.iloc[position]["topic_list"]

        # Content similarity
        content_score = cosine_sim[talk_position][position]

        # Topic similarity
        topic_score = calculate_topic_similarity(
            input_topics,
            candidate_topics
        )

        # Combine both scores
        hybrid_score = (0.8 * content_score) + (0.2 * topic_score)

        recommendations.append({
            "Title": df.iloc[position]["title"],
            "Speaker": df.iloc[position]["speaker_1"],
            "Content Similarity": round(content_score, 4),
            "Topic Similarity": round(topic_score, 4),
            "Hybrid Score": round(hybrid_score, 4),
            "URL": df.iloc[position]["url"]
        })

    # Sort by hybrid score
    recommendations = sorted(
        recommendations,
        key=lambda x: x["Hybrid Score"],
        reverse=True
    )

    return pd.DataFrame(recommendations[:top_n])


print("Hybrid recommendation function created successfully!")
```

```python
# What this step does: Both models receive exactly the same 500 input talks and the same 50 candidate talks per input. Only their ranking methods differ.
k = 5

# Use the same 500-talk sample for both models
sample_positions = np.random.default_rng(42).choice(
    len(df), size=500, replace=False
)

tfidf_results = []
hybrid_results = []

for position in sample_positions:

    relevant_ids = df.iloc[position]["valid_related_ids"]

    if len(relevant_ids) == 0:
        continue

    input_topics = set(df.iloc[position]["topic_list"])

    # Select the same 50 candidate talks for both models
    candidate_positions = [
        p for p in np.argsort(-cosine_sim[position])
        if p != position
    ][:50]

    candidate_scores = []

    for candidate_position in candidate_positions:

        candidate_topics = set(
            df.iloc[candidate_position]["topic_list"]
        )

        union = input_topics.union(candidate_topics)

        topic_score = (
            len(input_topics.intersection(candidate_topics)) / len(union)
            if union else 0
        )

        content_score = cosine_sim[position][candidate_position]

        candidate_scores.append(
            (candidate_position, content_score, topic_score)
        )

    # TF-IDF ranking
    tfidf_ranked = sorted(
        candidate_scores,
        key=lambda x: x[1],
        reverse=True
    )

    # Hybrid ranking
    hybrid_ranked = sorted(
        candidate_scores,
        key=lambda x: 0.8 * x[1] + 0.2 * x[2],
        reverse=True
    )

    tfidf_results.append([
        int(df.iloc[p]["talk_id"])
        for p, _, _ in tfidf_ranked[:k]
    ])

    hybrid_results.append([
        int(df.iloc[p]["talk_id"])
        for p, _, _ in hybrid_ranked[:k]
    ])

print("Fair comparison candidate lists generated!")
print("Evaluated talks:", len(tfidf_results))
print("Candidates per talk: 50")
```

Calculate all four evaluation metrics for both models: Precision@5, Recall@5, NDCG@5 and MAP@5.


What these metrics measure:
- Precision@5: How many of the top 5 recommendations are relevant?
- Recall@5: How many relevant talks did the model successfully recommend?
- NDCG@5: Does the model place relevant recommendations near the top?
- MAP@5: How accurately does the model rank relevant recommendations?

### Evaluation Results and Model Comparison

Both TF-IDF and Hybrid recommendation models were evaluated using the same 500-talk sample and the same 50 candidate recommendations per talk.

The Hybrid model achieved slightly better results across all four evaluation metrics:

- Precision@5: 0.3828
- Recall@5: 0.3216
- NDCG@5: 0.4247
- MAP@5: 0.3032

The TF-IDF model achieved Precision@5 of 0.3704, Recall@5 of 0.3113, NDCG@5 of 0.4148, and MAP@5 of 0.2958.

Therefore, the Hybrid model is selected as the preferred approach because combining textual similarity with topic similarity provides a modest improvement in recommendation quality.

Note: The `related_talks` field was used as proxy relevance data because actual user ratings and viewing history were unavailable.

```python

k = 5

def calculate_metrics(recommended_ids, relevant_ids):

    relevant_ids = set(relevant_ids)

    # Precision@5 and Recall@5
    hits = [
        1 if talk_id in relevant_ids else 0
        for talk_id in recommended_ids
    ]

    precision = sum(hits) / k
    recall = sum(hits) / len(relevant_ids)

    # DCG@5
    dcg = sum(
        hits[i] / np.log2(i + 2)
        for i in range(k)
    )

    # Ideal DCG@5
    ideal_hits = min(len(relevant_ids), k)

    idcg = sum(
        1 / np.log2(i + 2)
        for i in range(ideal_hits)
    )

    ndcg = dcg / idcg if idcg > 0 else 0

    # Average Precision@5
    precision_sum = 0
    relevant_count = 0

    for i in range(k):
        if hits[i] == 1:
            relevant_count += 1
            precision_sum += relevant_count / (i + 1)

    ap = precision_sum / min(len(relevant_ids), k)

    return precision, recall, ndcg, ap


tfidf_metrics = []
hybrid_metrics = []

for i, position in enumerate(sample_positions):

    relevant_ids = df.iloc[position]["valid_related_ids"]

    if len(relevant_ids) == 0:
        continue

    tfidf_metrics.append(
        calculate_metrics(tfidf_results[i], relevant_ids)
    )

    hybrid_metrics.append(
        calculate_metrics(hybrid_results[i], relevant_ids)
    )


# Calculate average scores
tfidf_average = np.mean(tfidf_metrics, axis=0)
hybrid_average = np.mean(hybrid_metrics, axis=0)

# Display comparison table
comparison_df = pd.DataFrame({
    "Metric": ["Precision@5", "Recall@5", "NDCG@5", "MAP@5"],
    "TF-IDF": tfidf_average,
    "Hybrid": hybrid_average
})

comparison_df[["TF-IDF", "Hybrid"]] = comparison_df[
    ["TF-IDF", "Hybrid"]
].round(4)

print("MODEL EVALUATION COMPARISON")
print("Evaluated talks:", len(tfidf_metrics))
display(comparison_df)


# Metric names
metrics = comparison_df["Metric"]

# Model scores
tfidf_scores = comparison_df["TF-IDF"]
hybrid_scores = comparison_df["Hybrid"]

# Bar positions
x = np.arange(len(metrics))
width = 0.35

# Create chart
plt.figure(figsize=(10, 5))

plt.bar(x - width/2, tfidf_scores, width, label="TF-IDF")
plt.bar(x + width/2, hybrid_scores, width, label="Hybrid")

# Chart details
plt.xlabel("Evaluation Metrics")
plt.ylabel("Score")
plt.title("TF-IDF vs Hybrid Model Evaluation")
plt.xticks(x, metrics)
plt.ylim(0, 0.5)
plt.legend()
plt.grid(axis="y", linestyle="--", alpha=0.5)

plt.tight_layout()
plt.show()
```

### Comparison of Content-Based and Collaborative Filtering

**1. Content-Based Filtering**

Content-Based Filtering recommends TED Talks based on similarities in their textual content, descriptions, transcripts, and topics.

Advantages:
- Does not require user ratings or interaction history.
- Can recommend talks to new users.
- Provides recommendations based on content similarity.
- Easy to interpret and explain.

Limitations:
- May recommend talks that are too similar to previously selected talks.
- Recommendation quality depends on the available content and metadata.
- May provide fewer unexpected or diverse recommendations.

**2. Collaborative Filtering**

Collaborative Filtering recommends items based on the preferences and behavior of multiple users.

Advantages:
- Can identify recommendations based on similar user interests.
- Can discover items with different content but similar user appeal.
- Can provide personalized recommendations.

Limitations:
- Requires user ratings, clicks, or viewing history.
- Faces cold-start problems for new users and items.
- Performance can be affected by sparse interaction data.

**3. Approach Selected**

Collaborative Filtering was not implemented because the available TED Talk dataset does not contain user-level ratings or viewing history.

Therefore, Content-Based Filtering using TF-IDF and topic similarity was selected. The Hybrid model combines both approaches to improve recommendation quality while remaining interpretable.

### Final Model Selection and Justification

After evaluating both TF-IDF and Hybrid recommendation models using Precision@5, Recall@5, NDCG@5, and MAP@5, the Hybrid model was selected as the final recommendation approach.

**Reasons for selecting the Hybrid model:**

1. **Better Performance:** The Hybrid model achieved higher scores across all four evaluation metrics.
2. **Improved Relevance:** Combining textual similarity and topic similarity helps identify more relevant TED Talks.
3. **Better Ranking:** NDCG@5 and MAP@5 indicate a modest improvement in ranking relevant recommendations.
4. **Interpretability:** The model combines two understandable similarity measures, making its recommendations easier to explain.
5. **No User History Required:** It works with the available TED Talk content and metadata without requiring user ratings or viewing history.

**Final Model Configuration:**

- Text similarity method: TF-IDF
- Topic similarity method: Jaccard Similarity
- Content similarity weight: 80%
- Topic similarity weight: 20%
- Number of recommendations: Top 5

**Conclusion:**

The Hybrid recommendation model was selected because it provides a balanced and interpretable recommendation approach with slightly better offline evaluation results than the standalone TF-IDF model.
