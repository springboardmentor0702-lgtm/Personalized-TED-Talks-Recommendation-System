```python
## Text Analysis
display(HTML("<h2 style='text-align:center;'>Text Analysis</h2>"))
#word count
df["title_word_count"] = df["title"].str.split().str.len()
df["description_word_count"] = df["description"].str.split().str.len()
df["transcript_word_count"] = df["transcript"].str.split().str.len()
display(df[[
    "title_word_count",
    "description_word_count",
    "transcript_word_count"
]].describe())

#find unusual short titles
display("unusual short titles",
    df[df["title_word_count"] <= 2][
        ["title","speaker_1"]
    ]
)

display(HTML("<h2 style='text-align:center;'>Lengthy Transcripts</h2>"))
#find extremely long talk transcripts
display("extremely long transcripts",
    df.nlargest(
        10,
        "transcript_word_count"
    )[["title","transcript_word_count"]]
)

display(pd.DataFrame({
    "Column":[
        "Title",
        "Description",
        "Transcript",
        "About Speakers"
    ],
    "Missing":[
        df["title"].isna().sum(),
        df["description"].isna().sum(),
        df["transcript"].isna().sum(),
        df["about_speakers"].isna().sum()
    ]
}))

display(HTML("<h2 style='text-align:center;'>Transcript Quality Check</h2>"))
#transcript quality- check if they r usable
display(df["transcript_word_count"].describe())


display(HTML("<h2 style='text-align:center;'>Related Talk Analysis</h2>"))
#Related talks analysis
display("Related Talk Analysis",df["related_talks"].head())
#count related talks
df["related_count"] = (
    df["related_talks"]
      .fillna("")
      .str.count(",") + 1
)
#check if ID's exist
talk_ids = set(df["talk_id"])
#verify ID's
def check_ids(text):
    if pd.isna(text):
        return True

    ids = text.replace("[","").replace("]","").split(",")
    return all(
        int(i.strip()) in talk_ids
        for i in ids
        if i.strip().isdigit()
    )
df["related_valid"] = df["related_talks"].apply(check_ids)
```

```python
#Data Cleaning Decisions
text_columns = ["title", "topics", "description", "transcript"]
for col in text_columns:
    df[col] = df[col].fillna("").astype(str)

#cleaning code
def clean_text(text):
    text = text.lower()
    text = re.sub(r"http\S+|www\.\S+", " ", text)
    text = re.sub(r"[^a-z0-9\s]", " ", text)
    text = re.sub(r"\s+", " ", text)
    return text.strip()

df["occupations"] = df["occupations"].fillna("Unknown")
df["about_speakers"] = df["about_speakers"].fillna("")
df["all_speakers"] = df["all_speakers"].fillna(df["speaker_1"])

#Standardize text
df["title"] = (
    df["title"]
      .str.lower()
      .str.strip()
)
df["description"] = (
    df["description"]
      .str.lower()
      .str.strip()
)
df["transcript"] = (
    df["transcript"]
      .str.lower()
      .str.strip()
)
df["topics"] = (
    df["topics"]
      .str.lower()
      .str.strip()
)

#create combined text
df["combined_text"] = (
    df["title"].fillna("") + " " +
    df["topics"].fillna("") + " " +
    df["description"].fillna("") + " " +
    df["transcript"].fillna("")
)

display(df[["title", "combined_text"]].head(3))
```

```python
#TF-IDF

# re — helps clean unnecessary characters and extra spaces.
# TfidfVectorizer — converts text into numerical vectors.
# cosine_similarity — measures how similar two TED Talks are.
import re
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

display(HTML("<h2 style='text-align:center;'>TF- IDF</h2>"))
tfidf = TfidfVectorizer(
    stop_words="english",
    max_features=10000,
    ngram_range=(1, 2),
    min_df=2
)

tfidf_matrix = tfidf.fit_transform(df["combined_text"])

print("TF-IDF matrix shape:", tfidf_matrix.shape)


#Recommendation Engine

def recommend_talk(title, top_n=5):
    matches = df.index[
        df["title"].str.casefold() == title.casefold()
    ].tolist()

    if not matches:
        return f"No TED Talk found with title: {title}"

    talk_index = matches[0]

    similarities = cosine_similarity(
        tfidf_matrix[talk_index],
        tfidf_matrix
    ).flatten()

    similarities[talk_index] = -1

    top_indices = similarities.argsort()[-top_n:][::-1]

    result = df.iloc[top_indices][
        ["title", "speaker_1", "views", "url"]
    ].copy()

    result["similarity_score"] = similarities[top_indices].round(3)

    return result.reset_index(drop=True)
```

```python
#Testing recommendation engine
recommend_talk("Do schools kill creativity?", top_n=5)

recommend_talk(df["title"].iloc[100], top_n=5)

print("Total TED Talks:", len(df))
print("TF-IDF documents:", tfidf_matrix.shape[0])
print("Vocabulary size:", tfidf_matrix.shape[1])
print("Combined text missing:", df["combined_text"].isna().sum())
```
