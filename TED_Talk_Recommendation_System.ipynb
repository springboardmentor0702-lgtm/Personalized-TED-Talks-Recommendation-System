```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline

# Load dataset using Python engine to handle unclosed quotes robustly
df = pd.read_csv("/content/ted_talks_en.csv", engine='python', on_bad_lines='skip')

# STEP 2: Basic inspection
display("Shape:", df.shape)
display(" First 5 rows:", df.head())
display(" Last 5 rows:", df.tail())
display(" Columns:",df.columns.tolist())
display(" Data types:",df.dtypes)
display(" Dataset information:",df.info())

# STEP 3: Statistical summary
display("Numeric summary:",df.describe())
display(" View summary:",df["views"].describe())
display(" Duration summary:",df["duration"].describe())
display("Complete summary:",df.describe(include="all"))
display("Object summary:",df.describe(include="object"))

# STEP 4: Missing values
missing = df.isnull().sum().sort_values(ascending=False)
display("Missing values:",missing)
display("Missing percentage:")
display((df.isnull().mean() * 100).round(2).sort_values(ascending=False))

# visualize missing values using heatmap
plt.figure(figsize=(12,6))
sns.heatmap(df.isnull(),
            cbar=False,
            yticklabels=False)
plt.title("Missing Values in TED Talks Dataset")
plt.show()

# STEP 5: Duplicate checks
display("Full duplicate rows:", df.duplicated().sum())
display("Duplicate talk_id:", df["talk_id"].duplicated().sum())

#Check unique values
display(" count of i=unique categorical values",df.nunique().sort_values())
display("count of unique speakers:",df["speaker_1"].unique())
display("count of unique events:",df["event"].unique())
display("count of unique native languages:",df["native_lang"].unique())
display("count of unique talk_id:",df["talk_id"].unique())

# STEP 6: Numeric sanity checks
display("Comments range:", df["comments"].min(), "to", df["comments"].max())
display("Duration range:", df["duration"].min(), "to", df["duration"].max())
display("Zero-view talks:", (df["views"] == 0).sum())
display("Views range:", df["views"].min(), "to", df["views"].max())
plt.figure(figsize=(10,6))
sns.histplot(df["views"], bins=50, kde=True)
plt.title("Distribution of TED Talk Views")
plt.xlabel("Views")
plt.ylabel("Number of Talks")
plt.show()

#log transform views
df["log_views"] = np.log1p(df["views"])
plt.figure(figsize=(10,6))
sns.histplot(df["log_views"], bins=50, kde=True)
plt.title("Distribution of Log-Transformed Views")
plt.xlabel("log(1 + views)")
plt.ylabel("Number of Talks")
plt.show()

#most viewed TED Talks:
top_talks = df.sort_values(
    by="views",
    ascending=False
)
display(" most viewed TED Talks:",top_talks[["title", "speaker_1", "views"]].head(10))

#Analyze the speakers
display("Speakers analysis:",df["speaker_1"].value_counts().head(15))
top_speakers = df["speaker_1"].value_counts().head(15)
plt.figure(figsize=(10,6))
top_speakers.sort_values().plot(kind="barh")
plt.title("Top 15 Speakers by Number of TED Talks")
plt.xlabel("Number of Talks")
plt.ylabel("Speaker")
plt.show()

#Analyze TED Events
display("Events analysis:",df["event"].value_counts().head(15))
top_events = df["event"].value_counts().head(15)
plt.figure(figsize=(10,6))
top_events.sort_values().plot(kind="barh")
plt.title("Top 15 TED Events")
plt.xlabel("Number of Talks")
plt.ylabel("Event")
plt.show()

#Analyze native languages
display("Native languages analysis:",df["native_lang"].value_counts().head(5))
plt.figure(figsize=(10,5))
df["native_lang"].value_counts().plot(kind="bar")
plt.title("TED Talks by Native Language")
plt.xlabel("Language")
plt.ylabel("Number of Talks")
plt.xticks(rotation=45)
plt.show()

# STEP 7: Date conversion
df["recorded_date"] = pd.to_datetime(df["recorded_date"], errors="coerce")
df["published_date"] = pd.to_datetime(df["published_date"], errors="coerce")
display("Recorded date range:", df["recorded_date"].min(), "to", df["recorded_date"].max())
display("Published date range:", df["published_date"].min(), "to", df["published_date"].max())

# STEP 8: Derived features for EDA
df["recorded_year"] = df["recorded_date"].dt.year
df["log_views"] = np.log1p(df["views"])
df["log_comments"] = np.log1p(df["comments"].fillna(0))

#Analyze duration
df["duration_minutes"] = df["duration"].astype(float) / 60
display(df["duration_minutes"].describe())
plt.figure(figsize=(10,6))
sns.histplot(
    df["duration_minutes"],
    bins=40,
    kde=True
)
plt.title("Distribution of TED Talk Duration")
plt.xlabel("Duration in Minutes")
plt.ylabel("Number of Talks")
plt.show()

#Analyze publication yr
df["published_year"] = df["published_date"].dt.year
year_counts = df["published_year"].value_counts().sort_index()
print("year_counts",year_counts.head())
plt.figure(figsize=(12,5))
year_counts.plot(
    kind="line",
    marker="o"
)
plt.title("TED Talks Published by Year")
plt.xlabel("Year")
plt.ylabel("Number of Talks")
plt.grid(True)
plt.show()

#shortest nd longest talks
print(df.loc[
    df["duration"].idxmin(),
    ["title", "duration"]
],
df.loc[
    df["duration"].idxmax(),
    ["title", "duration"]
])

#Analyze comments
display(df["comments"].describe())
plt.figure(figsize=(10,6))

sns.histplot(
    df["comments"].dropna(),
    bins=50,
    kde=True
)

plt.title("Distribution of TED Talk Comments")
plt.xlabel("Comments")
plt.ylabel("Number of Talks")

plt.show()

#views vs. comments
plt.figure(figsize=(10,6))
sns.scatterplot(
    data=df,
    x="comments",
    y="log_views",
    alpha=0.5
)
plt.title("Comments vs TED Talk Views")
plt.xlabel("Comments")
plt.ylabel("log(1 + Views)")
plt.show()

#views vs duration
plt.figure(figsize=(10,6))
sns.scatterplot(
    data=df,
    x="duration_minutes",
    y="log_views",
    alpha=0.5
)
plt.title("Talk Duration vs Views")
plt.xlabel("Duration (Minutes)")
plt.ylabel("log(1 + Views)")
plt.show()

#correlation analysis
numeric_cols = [
    "talk_id",
    "views",
    "comments",
    "duration"
]
corr = df[numeric_cols].corr()
display(corr)

plt.figure(figsize=(8,5))
sns.heatmap(
    corr,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)
plt.title("Correlation Matrix")
plt.show()

#topic analysis
display(" topics :",df["topics"].head(10))
display(" unique topics :",df["topics"].nunique())

#description
display(" description :",df["description"].head())
#checking for missing values
display(" description missing sum :",df["description"].isnull().sum())

#Exploring transcripts
display(" transcripts :",df["transcript"].head())
display(" description missing sum :",df["transcript"].isnull().sum())

#checking transcript length
df["transcript_length"] = df["transcript"].str.len()
df["transcript_length"].describe()

plt.figure(figsize=(10,6))
sns.histplot(
    df["transcript_length"],
    bins=50
)
plt.title("Distribution of Transcript Length")
plt.xlabel("Characters")
plt.ylabel("Number of Talks")
plt.show()

#analyzing speaker occupation
display(df["occupations"].value_counts().head(10))

#checking zero values
display("Zero views:", (df["views"] == 0).sum())
display("Zero comments:", (df["comments"] == 0).sum())
display("Zero duration:", (df["duration" ] == 0).sum())

#outlier analysis for views
plt.figure(figsize=(10,4))
sns.boxplot(x=df["views"])
plt.title("Boxplot of TED Talk Views")
plt.show()

eda_summary = pd.DataFrame({
    "Metric": [
        "Number of rows",
        "Number of columns",
        "Duplicate rows",
        "Duplicate talk IDs",
        "Missing comments",
        "Missing occupations",
        "Missing speaker information",
        "Unique speakers",
        "Unique events",
        "Unique native languages"
    ],
    "Value": [
        df.shape[0],
        df.shape[1],
        df.duplicated().sum(),
        df["talk_id"].duplicated().sum(),
        df["comments"].isnull().sum(),
        df["occupations"].isnull().sum(),
        df["about_speakers"].isnull().sum(),
        df["speaker_1"].nunique(),
        df["event"].nunique(),
        df["native_lang"].nunique()
    ]
})

print(eda_summary)

df.to_csv(
    "ted_talks_EDA_ready.csv",
    index=False
)
```
